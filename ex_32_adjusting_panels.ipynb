```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl

%matplotlib inline
```

Panel adjustments are best made using grid stack, which will naturally push you towards designing aesthetically pleasing well organized figures. In some circumstances, it can be useful to manually tweak axes positions, such as for the color bar.

# Example data

```python
x = np.linspace(0, 10, 100)
y = np.linspace(0, 10, 100)
X, Y = np.meshgrid(x, y)
Z = np.abs(np.sin(X) * np.cos(Y))**2
```

# Plot

```python
fig = plt.figure(figsize=(6.75, 6.75/2))

gs = mpl.gridspec.GridSpec(2, 2,  height_ratios=[0.04, 1], hspace=0.4)

for i in range(2):
    ax = plt.subplot(gs[1, i])
    art = ax.pcolormesh(X, Y, Z, vmin=0, vmax=1)

    ax.xaxis.set_minor_locator(mpl.ticker.AutoMinorLocator(2))
    ax.yaxis.set_minor_locator(mpl.ticker.AutoMinorLocator(2))

cax = plt.subplot(gs[0, :])
pos = cax.get_position()
new_width = pos.width * 0.5
new_x0 = pos.x0 + (pos.width - new_width) / 2
cax.set_position([new_x0, pos.y0, new_width, pos.height])

plt.colorbar(art, cax=cax, orientation='horizontal',
             label='Intensity (Arb. units)')
fig.savefig('ex_32_adjusting_panels.pdf', dpi=1200)
```
